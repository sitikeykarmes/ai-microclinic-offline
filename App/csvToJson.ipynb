```python
import csv
import json

def csv_to_json(csv_file, json_file):
    data = {}
    with open(csv_file, 'r', encoding='utf-8') as f:
        reader = csv.reader(f)
        next(reader)  # Skip header
        for row in reader:
            key = row[0].strip().lower()
            value = row[1].strip()
            data[key] = value
    with open(json_file, 'w', encoding='utf-8') as f:
        json.dump(data, f, indent=2, ensure_ascii=False)

csv_to_json('../../ai-microclinic-offline-dataset/Disease Symptom Prediction/symptom_Description.csv', '../assets/disease_descriptions.json')

csv_to_json('../../ai-microclinic-offline-dataset/Disease Symptom Prediction/Symptom-severity.csv', '../assets/disease_severity.json')
```

```python
import csv
import json

def csv_to_precautions_json(csv_file_path, json_file_path):
    precautions_dict = {}

    with open(csv_file_path, newline='', encoding='utf-8') as csvfile:
        reader = csv.DictReader(csvfile)
        for row in reader:
            disease = row['Disease'].strip().lower()

            # Collect all non-empty precautions into a list
            precautions = []
            for key in row:
                if key.lower().startswith('precaution') and row[key].strip():
                    precautions.append(row[key].strip())

            precautions_dict[disease] = precautions

    with open(json_file_path, 'w', encoding='utf-8') as jsonfile:
        json.dump(precautions_dict, jsonfile, indent=2, ensure_ascii=False)

    print(f"✅ JSON file created successfully at: {json_file_path}")

# Usage:
csv_to_precautions_json(
    '../../ai-microclinic-offline-dataset/Disease Symptom Prediction/symptom_precaution.csv',
    '../assets/disease_precautions.json'
)
```
